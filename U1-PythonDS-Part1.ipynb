# Listas


Una lista en Python es una estructura de datos mutable y ordenada que permite almacenar una colección de elementos (de cualquier tipo: enteros, cadenas, objetos, etc.) en una sola variable. Se define usando corchetes [] y los elementos se separan por comas.

Características Principales:
- Mutable: Los elementos de una lista se pueden modificar después de su creación (agregar, eliminar, cambiar).
- Ordenada: Los elementos tienen un orden definido, accesible mediante índices (empezando en 0).
- Heterogénea: Puede contener elementos de diferentes tipos de datos (ej. [1, "hola", 3.14]).
- Indexable: Se accede a los elementos mediante índices (positivos o negativos).
- Admite duplicados: Permite elementos repetidos.
- Dinámica: Puede crecer o reducirse en tamaño usando métodos como append(), remove(), etc.
- Iterable: Se puede recorrer con bucles (ej. for).

# Tuplas

Una tupla en Python es una estructura de datos inmutable y ordenada que almacena una colección de elementos de cualquier tipo (enteros, cadenas, objetos, etc.). Se define usando paréntesis () y los elementos se separan por comas.

Características Principales:
- Inmutable: No se pueden modificar (agregar, eliminar o cambiar) los elementos después de crear la tupla.
- Ordenada: Los elementos tienen un orden definido, accesible mediante índices (desde 0).
- Heterogénea: Puede contener elementos de diferentes tipos (ej. (1, "hola", 3.14)).
- Indexable: Se accede a los elementos con índices (positivos o negativos).
- Admite duplicados: Permite elementos repetidos.
- Ligera y eficiente: Las tuplas consumen menos memoria que las listas y son más rápidas en ciertas operaciones.
- Iterable: Se puede recorrer con bucles (ej. for)..

Métodos Principales
- count(x): Cuenta cuántas veces aparece x en la tupla.
- len(x): Devuelve la longitud de la tupla.
- index(x[, start[, end]]): Devuelve el índice de la primera aparición de x en la tupla, opcionalmente dentro de un rango [start, end).
  
Usos Comunes
- Almacenar datos que no deben modificarse (ej. coordenadas, registros fijos).
- Como claves en diccionarios (a diferencia de las listas, que no son hashable).
- Devolver múltiples valores desde una función.

# Diccionarios

Un **diccionario** en Python es una estructura de datos **mutable** y desordenada (hasta Python 3.6; ordenada por inserción desde Python 3.7) que almacena pares **clave-valor**. Se define usando llaves `{}` o la función `dict()`, donde cada clave es única y está asociada a un valor.

### Características Principales
1. **Mutable**: Los elementos (pares clave-valor) pueden modificarse, agregarse o eliminarse.
2. **Claves únicas**: Las claves deben ser inmutables (ej. cadenas, números, tuplas) y no se permiten duplicados.
3. **Valores diversos**: Los valores pueden ser de cualquier tipo (números, cadenas, listas, etc.).
4. **Acceso por clave**: Los elementos se acceden usando la clave en lugar de un índice (ej. `diccionario[clave]`).
5. **Desordenado (hasta Python 3.6)**: Desde Python 3.7, los diccionarios mantienen el orden de inserción.
6. **Iterable**: Se puede recorrer con bucles para acceder a claves, valores o pares.
7. **Dinámico**: Puede crecer o reducirse con métodos como `pop()` o `update()`.

### Métodos Comunes
- `get(clave)`: Obtiene el valor de una clave, devuelve `None` si no existe (o un valor por defecto).
- `pop(clave)`: Elimina y devuelve el valor de la clave.
- `update()`: Actualiza o agrega pares clave-valor.
- `keys()`: Devuelve las claves.
- `values()`: Devuelve los valores.
- `items()`: Devuelve los pares clave-valor.


Los diccionarios son ideales para almacenar datos estructurados con acceso rápido por clave, como configuraciones o registros.

# Funciones

Sintaxis de una función de Python

En Python, el paso de parámetros a una función no se realiza estrictamente por referencia ni por valor como en otros lenguajes (por ejemplo, C++). En su lugar, Python utiliza un mecanismo conocido como **"paso por asignación"** (o call by object reference). Esto significa que los parámetros se pasan como referencias a los objetos, pero el comportamiento depende de si el objeto es mutable o inmutable.

Tipos de objetos:
- Objetos Mutables: Objetos como listas, diccionarios y conjuntos pueden modificarse directamente (en su lugar) porque son mutables. Si pasas un objeto mutable a una función, los cambios realizados dentro de la función afectan al objeto original fuera de ella.
- Objetos Inmutables: Objetos como enteros, flotantes, cadenas y tuplas no pueden modificarse después de crearse. Cualquier operación que parezca modificar un objeto inmutable crea un nuevo objeto, dejando el original intacto.

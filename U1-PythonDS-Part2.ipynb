# Gestión de archivos de Texto y Binarios

La **gestión de archivos** en Python permite trabajar con archivos de **texto** y **binarios**, dos tipos fundamentales con características y usos distintos. 

En ambos casos siempre se procede a abrir (**open**) el archivo y cerrarlo (con **close()** pero mejor usar **with** para cierre automático). 

Los modos de apertura son:
- 'r': Lectura (por defecto).
- 'w': Escritura (crea o sobrescribe el archivo).
- 'a': Añadir (agrega datos al final).
- 'rb', 'wb': Modos binarios para lectura/escritura.

## Diferencias entre Archivos de Texto y Binarios

- **Archivos de Texto**:
  - Contienen datos legibles por humanos, como cadenas de caracteres (ej. `.txt`, `.csv`, `.json`).
  - Almacenan texto codificado (generalmente en UTF-8).
  - Modos de apertura: `'r'` (lectura), `'w'` (escritura), `'a'` (añadir).
  - Ejemplo: Un archivo `.txt` con una lista de nombres.

- **Archivos Binarios**:
  - Contienen datos no legibles directamente, como imágenes, audio o ejecutables (ej. `.jpg`, `.mp3`, `.exe`).
  - Almacenan bytes crudos, sin interpretación textual.
  - Modos de apertura: `'rb'` (lectura binaria), `'wb'` (escritura binaria), `'ab'` (añadir binario).
  - Ejemplo: Un archivo `.png` con datos de imagen.

## Métodos Básicos

### Para Archivos de Texto
- **`open()`**: Abre un archivo de texto.
- **`read()`**: Lee todo el contenido como una cadena.
- **`readline()`**: Lee una línea a la vez.
- **`readlines()`**: Lee todas las líneas en una lista.
- **`write(texto)`**: Escribe una cadena en el archivo.
- **`writelines(lista)`**: Escribe una lista de cadenas.

```python
# Ejemplo de texto
with open('texto.txt', 'w', encoding='utf-8') as f:
    f.write('Hola\nMundo')
with open('texto.txt', 'r', encoding='utf-8') as f:
    print(f.read())  # Salida: Hola\nMundo
```

### Para Archivos Binarios
- **`open(,'*b')`**: Abre un archivo binario.
- **`read()`**: Lee todos los bytes como un objeto `bytes`.
- **`read(n)`**: Lee `n` bytes.
- **`write(bytes)`**: Escribe un objeto `bytes` en el archivo.

```python
# Ejemplo binario
with open('datos.bin', 'wb') as f:
    f.write(b'\x00\x01\x02')
with open('datos.bin', 'rb') as f:
    print(f.read())  # Salida: b'\x00\x01\x02'
```

## Notas
- Usa `encoding='utf-8'` para archivos de texto para evitar problemas de codificación.
- Usa el bloque `with` para garantizar que los archivos se cierren correctamente.
- Los archivos binarios requieren datos en formato `bytes` (usa `b''` o funciones como `bytearray()`).
- En ciencia de datos es muy común utilizar ficheros binarios en Python, aunque normalmente no se manipulan directamente con open("archivo", "rb") salvo en casos muy específicos. En su lugar, se usan formatos binarios especializados, diseñados para almacenar y recuperar datos de forma eficiente.

## Ficheros CSV
Los archivos CSV son un estándar de facto para datasets en ciencia de datos, especialmente en etapas iniciales o para proyectos con datos tabulares de tamaño manejablelos archivos de datos en formato CSV. Por ello se utilizan ampliamente como una opción estándar para almacenar y trabajar con datasets.
```
nombre,edad,ciudad
Ana,25,"San José"
Juan,30,Madrid
María,28,"Nueva York"
```
En Python hay ubiblioteca estándar llamada `csv`diseñada específicamente para la gestión de archivos en formato CSV. Esta biblioteca proporciona herramientas para leer, escribir y manipular archivos CSV de manera eficiente y robusta, manejando detalles como delimitadores, comillas y saltos de línea de forma automática.
### Métodos

- **`csv.reader()`**: Lee un archivo CSV y devuelve un iterador de filas como listas.
- **`csv.writer()`**: Escribe datos en un archivo CSV, fila por fila.
- **`csv.DictReader()`**: Lee un archivo CSV y devuelve un iterador de diccionarios, usando la primera fila como claves.
- **`csv.DictWriter()`**: Escribe datos en un archivo CSV usando diccionarios, con encabezados especificados.

# Gestión de errores y depuración

## Resumen de las Sentencias de Captura de Excepciones `try`, `except` y `finally`

Las **sentencias de captura de excepciones** en Python (`try`, `except` y `finally`) son herramientas esenciales para manejar errores durante la ejecución de un programa, permitiendo un control robusto y predecible del flujo de ejecución. A continuación, se ofrece un resumen conciso de cada una:

### `try`
- **Propósito**: Define un bloque de código donde se intentará ejecutar una operación que podría generar una excepción (error).
- **Uso**: Se utiliza para envolver el código que se desea monitorear por posibles errores.
- **Ejemplo**:
  ```python
  try:
      resultado = 10 / 0
  ```
  - Si ocurre un error (como `ZeroDivisionError`), el control pasa al bloque `except`.

### `except`
- **Propósito**: Captura y gestiona las excepciones que ocurren en el bloque `try`, permitiendo definir acciones específicas según el tipo de error.
- **Uso**: Puede especificar un tipo de excepción (ej. `ValueError`, `FileNotFoundError`) o manejarlos en general con una cláusula genérica.
- **Opciones**:
  - `except Exception as e`: Captura cualquier excepción y permite acceder al mensaje de error con `e`.
  - Múltiples `except`: Para manejar diferentes tipos de excepciones.
- **Ejemplo**:
  ```python
  try:
      resultado = 10 / 0
  except ZeroDivisionError:
      print("No se puede dividir por cero")
  except Exception as e:
      print(f"Ocurrió un error: {e}")
  ```
  - Salida: `No se puede dividir por cero`.

### `finally`
- **Propósito**: Define un bloque de código que se ejecuta siempre, independientemente de si se produjo una excepción o no, incluso después de un `return`.
- **Uso**: Ideal para liberar recursos (cerrar archivos, conexiones de red) o realizar limpieza.
- **Ejemplo**:
  ```python
  try:
      archivo = open('datos.txt', 'r')
      dato = archivo.read()
  except FileNotFoundError:
      print("Archivo no encontrado")
  finally:
      print("Cerrando archivo")
      archivo.close() if 'archivo' in locals() else None
  ```
  - Salida (si el archivo no existe): `Archivo no encontrado` y `Cerrando archivo`.

### Características Clave
- **Combinación**: `try` debe ir acompañado de al menos un `except` o `finally`, aunque lo más común es usar `try` con `except` y opcionalmente `finally`.
- **Jerarquía**: Si se especifican múltiples `except`, se evalúan en orden, y solo se ejecuta el primero que coincida.
- **Manejo de Recursos**: El uso de `finally` es especialmente útil con el contexto `with`, que automatiza la gestión de recursos (ej. cierre de archivos).
- **Excepciones Personalizadas**: Puedes definir tus propias excepciones con `raise`.

### Ejemplo Completo
```python
try:
    numero = int(input("Ingresa un número: "))
    resultado = 10 / numero
except ValueError:
    print("Entrada no válida, debe ser un número")
except ZeroDivisionError:
    print("No se puede dividir por cero")
finally:
    print("Ejecución finalizada")
```
- **Salida posible** (entrada `0`): `No se puede dividir por cero` y `Ejecución finalizada`.

### Resumen
- **`try`**: Intenta ejecutar código que podría fallar.
- **`except`**: Maneja errores específicos o generales.
- **`finally`**: Ejecuta código de limpieza siempre, sin importar el resultado.
Estas sentencias permiten crear programas más robustos, gestionando errores de forma controlada y asegurando la ejecución de tareas críticas.
